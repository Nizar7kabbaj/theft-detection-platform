```python
import pickle
from pathlib import Path
import numpy as np
from collections import defaultdict

ROOT     = Path("/kaggle/input/datasets/nizarkabbaj/poselift-pickle")
TEST_DIR = ROOT / "Test"
GT_DIR   = ROOT / "GT"

WINDOW = 30
STRIDE = 15

total_frames        = 0
total_anomaly_frames = 0
total_tracklets     = 0
total_windows       = 0
anomaly_windows     = 0
tracklet_lengths    = []
files_with_anomaly  = 0

for pkl_path in sorted(TEST_DIR.glob("*.pkl")):
    with open(pkl_path, "rb") as f:
        data = pickle.load(f)

    gt = np.load(GT_DIR / (pkl_path.stem + ".npy"))
    n_frames = len(data)

    total_frames += n_frames
    total_anomaly_frames += int(gt.sum())
    if gt.sum() > 0:
        files_with_anomaly += 1

    tracklets = defaultdict(list)
    for frame_idx in sorted(data.keys()):
        for pid in data[frame_idx]:
            tracklets[pid].append(frame_idx)

    for pid, frames in tracklets.items():
        if len(frames) < WINDOW:
            continue
        total_tracklets += 1
        tracklet_lengths.append(len(frames))

        for start in range(0, len(frames) - WINDOW + 1, STRIDE):
            window_frames = frames[start:start + WINDOW]
            window_label  = int(gt[window_frames].max())
            total_windows += 1
            anomaly_windows += window_label

print("=== DATASET SUMMARY (Test split, supervised demo mode) ===")
print(f"Files:                   47")
print(f"Files with anomaly:      {files_with_anomaly}")
print(f"Total frames:            {total_frames}")
print(f"Anomaly frames:          {total_anomaly_frames}  ({100*total_anomaly_frames/total_frames:.1f}%)")
print(f"Tracklets >= {WINDOW} frames:  {total_tracklets}")
if tracklet_lengths:
    print(f"Tracklet length min/median/max: "
          f"{min(tracklet_lengths)} / {int(np.median(tracklet_lengths))} / {max(tracklet_lengths)}")
print(f"\nWith WINDOW={WINDOW}, STRIDE={STRIDE}:")
print(f"Total windows:           {total_windows}")
print(f"Anomaly windows:         {anomaly_windows}  ({100*anomaly_windows/total_windows:.1f}%)" if total_windows else "")
print(f"Normal windows:          {total_windows - anomaly_windows}")
print(f"\nClass weight ratio (normal:anomaly): "
      f"{(total_windows - anomaly_windows) / max(anomaly_windows, 1):.2f}:1")
```

```python
import pickle
from pathlib import Path
from collections import defaultdict

import numpy as np
import torch
from torch.utils.data import Dataset

WINDOW       = 30
STRIDE       = 15
NUM_KP       = 17
FEAT_PER_KP  = 3
INPUT_SIZE   = NUM_KP * FEAT_PER_KP

ROOT     = Path("/kaggle/input/datasets/nizarkabbaj/poselift-pickle")
TEST_DIR = ROOT / "Test"
GT_DIR   = ROOT / "GT"


def normalize_keypoints(keypoints, bbox):
    kp = np.asarray(keypoints, dtype=np.float32).copy()
    x1, y1, x2, y2 = bbox

    cx = (x1 + x2) / 2.0
    cy = (y1 + y2) / 2.0
    half_w = max((x2 - x1) / 2.0, 1.0)
    half_h = max((y2 - y1) / 2.0, 1.0)

    kp[:, 0] = (kp[:, 0] - cx) / half_w
    kp[:, 1] = (kp[:, 1] - cy) / half_h

    bad = ~np.isfinite(kp).all(axis=1)
    kp[bad] = 0.0

    return kp


def build_windows_from_file(pkl_path, gt_path):
    with open(pkl_path, "rb") as f:
        data = pickle.load(f)
    gt = np.load(gt_path)

    tracklets = defaultdict(list)
    for frame_idx in sorted(data.keys()):
        for pid, person in data[frame_idx].items():
            bbox, keypoints = person[0], person[1]
            tracklets[pid].append((frame_idx, bbox, keypoints))

    samples = []
    for pid, entries in tracklets.items():
        if len(entries) < WINDOW:
            continue

        feats = np.stack([
            normalize_keypoints(kp, bb).flatten()
            for _, bb, kp in entries
        ], axis=0)

        frame_indices = [fi for fi, _, _ in entries]

        for start in range(0, len(entries) - WINDOW + 1, STRIDE):
            window_feats = feats[start:start + WINDOW]
            window_frames = frame_indices[start:start + WINDOW]
            label = int(gt[window_frames].max())
            samples.append((window_feats.astype(np.float32), label))

    return samples


class PoseLiftWindowDataset(Dataset):
    """
    A list-backed Dataset of fixed-length pose windows.
    Each item: (tensor (WINDOW, 51), label int).
    """
    def __init__(self, samples):
        self.samples = samples

    def __len__(self):
        return len(self.samples)

    def __getitem__(self, idx):
        x, y = self.samples[idx]
        return torch.from_numpy(x), torch.tensor(y, dtype=torch.long)


def build_splits(test_dir=TEST_DIR, gt_dir=GT_DIR, seed=42):
    """
    File-level split (no window leakage):
      70% train / 15% val / 15% test
    """
    files = sorted(test_dir.glob("*.pkl"))
    rng = np.random.default_rng(seed)
    indices = rng.permutation(len(files))

    n = len(files)
    n_train = int(round(0.70 * n))
    n_val   = int(round(0.15 * n))
    train_files = [files[i] for i in indices[:n_train]]
    val_files   = [files[i] for i in indices[n_train:n_train + n_val]]
    test_files  = [files[i] for i in indices[n_train + n_val:]]

    def windows_for(file_list):
        out = []
        for pkl in file_list:
            gt = gt_dir / (pkl.stem + ".npy")
            out.extend(build_windows_from_file(pkl, gt))
        return out

    return (
        windows_for(train_files),
        windows_for(val_files),
        windows_for(test_files),
        {"train_files": [f.name for f in train_files],
         "val_files":   [f.name for f in val_files],
         "test_files":  [f.name for f in test_files]},
    )


train_samples, val_samples, test_samples, file_split = build_splits()

train_ds = PoseLiftWindowDataset(train_samples)
val_ds   = PoseLiftWindowDataset(val_samples)
test_ds  = PoseLiftWindowDataset(test_samples)

def class_balance(samples):
    if not samples:
        return "0/0"
    labels = np.array([y for _, y in samples])
    return f"{int((labels == 0).sum())} normal / {int((labels == 1).sum())} anomaly"

print("=== DATASET SPLITS ===")
print(f"Train: {len(train_ds):4d} windows from {len(file_split['train_files'])} files  | {class_balance(train_samples)}")
print(f"Val:   {len(val_ds):4d} windows from {len(file_split['val_files'])} files  | {class_balance(val_samples)}")
print(f"Test:  {len(test_ds):4d} windows from {len(file_split['test_files'])} files  | {class_balance(test_samples)}")

x, y = train_ds[0]
print(f"\nSample tensor shape: {x.shape}  dtype: {x.dtype}")
print(f"Sample label: {y.item()}")
print(f"Feature range: x.min={x.min():.3f}, x.max={x.max():.3f}, x.mean={x.mean():.3f}")
print(f"Any NaN? {torch.isnan(x).any().item()}  Any inf? {torch.isinf(x).any().item()}")
```

```python
import torch
import torch.nn as nn

class ShoplifterLSTM(nn.Module):
    """
    Small LSTM classifier for pose-sequence binary classification.

    Input:  (batch, 30, 51)  — 30 frames of 51 features
    Output: (batch, 2)       — raw logits for [normal, anomaly]
    """
    def __init__(self, input_size=51, hidden_size=64, num_layers=2, dropout=0.3):
        super().__init__()
        self.hidden_size = hidden_size
        self.num_layers  = num_layers

        self.lstm = nn.LSTM(
            input_size=input_size,
            hidden_size=hidden_size,
            num_layers=num_layers,
            batch_first=True,
            dropout=dropout if num_layers > 1 else 0.0,
        )

        self.classifier = nn.Sequential(
            nn.Dropout(dropout),
            nn.Linear(hidden_size, 2),
        )

    def forward(self, x):
        _, (h_n, _) = self.lstm(x)
        last_hidden = h_n[-1]
        logits = self.classifier(last_hidden)
        return logits

DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
print(f"Device: {DEVICE}")
print(f"CUDA available: {torch.cuda.is_available()}")
if torch.cuda.is_available():
    print(f"GPU: {torch.cuda.get_device_name(0)}")

model = ShoplifterLSTM(input_size=51, hidden_size=64, num_layers=2, dropout=0.3).to(DEVICE)

n_params = sum(p.numel() for p in model.parameters() if p.requires_grad)
print(f"\nTrainable parameters: {n_params:,}")

x, y = train_ds[0]
x_batch = x.unsqueeze(0).to(DEVICE)
print(f"\nInput shape:  {x_batch.shape}")

with torch.no_grad():
    logits = model(x_batch)

print(f"Output shape: {logits.shape}  (should be (1, 2))")
print(f"Output logits: {logits.cpu().numpy()}")
print(f"Output probs:  {torch.softmax(logits, dim=1).cpu().numpy()}")
print(f"\nModel:\n{model}")
```

```python
import random
import numpy as np
import torch
import torch.nn as nn
from torch.utils.data import DataLoader
from sklearn.metrics import precision_recall_fscore_support, accuracy_score


SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

BATCH_SIZE   = 16
LR           = 1e-3
WEIGHT_DECAY = 1e-4
MAX_EPOCHS   = 60
PATIENCE     = 10

train_loader = DataLoader(train_ds, batch_size=BATCH_SIZE, shuffle=True,  num_workers=0)
val_loader   = DataLoader(val_ds,   batch_size=BATCH_SIZE, shuffle=False, num_workers=0)
test_loader  = DataLoader(test_ds,  batch_size=BATCH_SIZE, shuffle=False, num_workers=0)

train_labels = np.array([y for _, y in train_samples])
n_normal  = int((train_labels == 0).sum())
n_anomaly = int((train_labels == 1).sum())

w_normal  = (n_normal + n_anomaly) / (2.0 * n_normal)
w_anomaly = (n_normal + n_anomaly) / (2.0 * n_anomaly)
class_weights = torch.tensor([w_normal, w_anomaly], dtype=torch.float32, device=DEVICE)
print(f"Class weights: normal={w_normal:.3f}, anomaly={w_anomaly:.3f}")

criterion = nn.CrossEntropyLoss(weight=class_weights)

model = ShoplifterLSTM(input_size=51, hidden_size=64, num_layers=2, dropout=0.3).to(DEVICE)
optimizer = torch.optim.Adam(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)


def evaluate(model, loader):
    model.eval()
    total_loss = 0.0
    total_n    = 0
    all_preds  = []
    all_labels = []
    with torch.no_grad():
        for x, y in loader:
            x, y = x.to(DEVICE), y.to(DEVICE)
            logits = model(x)
            loss = criterion(logits, y)
            total_loss += loss.item() * x.size(0)
            total_n    += x.size(0)
            preds = logits.argmax(dim=1)
            all_preds.append(preds.cpu().numpy())
            all_labels.append(y.cpu().numpy())
    avg_loss = total_loss / max(total_n, 1)
    all_preds  = np.concatenate(all_preds)
    all_labels = np.concatenate(all_labels)
    acc = accuracy_score(all_labels, all_preds)
    p, r, f, _ = precision_recall_fscore_support(
        all_labels, all_preds, labels=[1], average=None, zero_division=0
    )
    return avg_loss, acc, float(p[0]), float(r[0]), float(f[0])


history = {"train_loss": [], "val_loss": [], "val_acc": [], "val_f1": []}
best_val_f1 = -1.0
best_state = None
epochs_no_improve = 0

print(f"\n{'Epoch':>5} | {'TrainLoss':>9} | {'ValLoss':>8} | {'ValAcc':>7} | {'ValPrec':>7} | {'ValRec':>7} | {'ValF1':>7}")
print("-" * 70)

for epoch in range(1, MAX_EPOCHS + 1):
    model.train()
    running_loss = 0.0
    running_n    = 0
    for x, y in train_loader:
        x, y = x.to(DEVICE), y.to(DEVICE)
        optimizer.zero_grad()
        logits = model(x)
        loss = criterion(logits, y)
        loss.backward()
        optimizer.step()
        running_loss += loss.item() * x.size(0)
        running_n    += x.size(0)
    train_loss = running_loss / max(running_n, 1)

    val_loss, val_acc, val_p, val_r, val_f1 = evaluate(model, val_loader)

    history["train_loss"].append(train_loss)
    history["val_loss"].append(val_loss)
    history["val_acc"].append(val_acc)
    history["val_f1"].append(val_f1)

    print(f"{epoch:>5} | {train_loss:>9.4f} | {val_loss:>8.4f} | {val_acc:>7.3f} | {val_p:>7.3f} | {val_r:>7.3f} | {val_f1:>7.3f}", end="")

    if val_f1 > best_val_f1:
        best_val_f1 = val_f1
        best_state  = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
        epochs_no_improve = 0
        print("  *")
    else:
        epochs_no_improve += 1
        print()
        if epochs_no_improve >= PATIENCE:
            print(f"\nEarly stopping at epoch {epoch}. Best val F1 = {best_val_f1:.3f}")
            break

if best_state is not None:
    model.load_state_dict(best_state)

test_loss, test_acc, test_p, test_r, test_f1 = evaluate(model, test_loader)
print(f"\n=== FINAL HELD-OUT TEST METRICS ===")
print(f"Test loss:      {test_loss:.4f}")
print(f"Test accuracy:  {test_acc:.3f}")
print(f"Test precision (anomaly): {test_p:.3f}")
print(f"Test recall    (anomaly): {test_r:.3f}")
print(f"Test F1        (anomaly): {test_f1:.3f}")
```

```python
import copy
from sklearn.model_selection import KFold
from sklearn.metrics import precision_recall_fscore_support, accuracy_score, confusion_matrix


N_FOLDS = 5

SEED = 42
random.seed(SEED)
np.random.seed(SEED)
torch.manual_seed(SEED)
torch.cuda.manual_seed_all(SEED)

BATCH_SIZE   = 16
LR           = 1e-3
WEIGHT_DECAY = 1e-4
MAX_EPOCHS   = 60
PATIENCE     = 10

all_files = sorted(TEST_DIR.glob("*.pkl"))
n_files = len(all_files)
print(f"Running {N_FOLDS}-fold CV on {n_files} files")
print(f"Each fold: ~{int(n_files * (N_FOLDS - 1) / N_FOLDS)} train / ~{int(n_files / N_FOLDS)} test files\n")

kf = KFold(n_splits=N_FOLDS, shuffle=True, random_state=SEED)

fold_results = []
fold_models  = []

def windows_for_files(file_list):
    out = []
    for pkl in file_list:
        gt = GT_DIR / (pkl.stem + ".npy")
        out.extend(build_windows_from_file(pkl, gt))
    return out


def evaluate_with_confusion(model, loader, criterion):
    model.eval()
    total_loss = 0.0
    total_n = 0
    all_preds, all_labels = [], []
    with torch.no_grad():
        for x, y in loader:
            x, y = x.to(DEVICE), y.to(DEVICE)
            logits = model(x)
            loss = criterion(logits, y)
            total_loss += loss.item() * x.size(0)
            total_n += x.size(0)
            preds = logits.argmax(dim=1)
            all_preds.append(preds.cpu().numpy())
            all_labels.append(y.cpu().numpy())
    avg_loss = total_loss / max(total_n, 1)
    all_preds = np.concatenate(all_preds)
    all_labels = np.concatenate(all_labels)
    acc = accuracy_score(all_labels, all_preds)
    p, r, f, _ = precision_recall_fscore_support(
        all_labels, all_preds, labels=[1], average=None, zero_division=0
    )
    cm = confusion_matrix(all_labels, all_preds, labels=[0, 1])
    return avg_loss, acc, float(p[0]), float(r[0]), float(f[0]), cm, all_preds, all_labels


for fold_idx, (train_idx, test_idx) in enumerate(kf.split(all_files), start=1):
    train_files = [all_files[i] for i in train_idx]
    test_files  = [all_files[i] for i in test_idx]

    rng = np.random.default_rng(SEED + fold_idx)
    perm = rng.permutation(len(train_files))
    n_val = max(3, len(train_files) // 6)
    val_files   = [train_files[i] for i in perm[:n_val]]
    train_files = [train_files[i] for i in perm[n_val:]]

    train_samples = windows_for_files(train_files)
    val_samples   = windows_for_files(val_files)
    test_samples  = windows_for_files(test_files)

    train_ds_f = PoseLiftWindowDataset(train_samples)
    val_ds_f   = PoseLiftWindowDataset(val_samples)
    test_ds_f  = PoseLiftWindowDataset(test_samples)

    train_loader = DataLoader(train_ds_f, batch_size=BATCH_SIZE, shuffle=True,  num_workers=0)
    val_loader   = DataLoader(val_ds_f,   batch_size=BATCH_SIZE, shuffle=False, num_workers=0)
    test_loader  = DataLoader(test_ds_f,  batch_size=BATCH_SIZE, shuffle=False, num_workers=0)

    train_labels = np.array([y for _, y in train_samples])
    n_normal  = max(int((train_labels == 0).sum()), 1)
    n_anomaly = max(int((train_labels == 1).sum()), 1)
    w_normal  = (n_normal + n_anomaly) / (2.0 * n_normal)
    w_anomaly = (n_normal + n_anomaly) / (2.0 * n_anomaly)
    class_weights = torch.tensor([w_normal, w_anomaly], dtype=torch.float32, device=DEVICE)
    criterion = nn.CrossEntropyLoss(weight=class_weights)

    torch.manual_seed(SEED + fold_idx)
    model = ShoplifterLSTM(input_size=51, hidden_size=64, num_layers=2, dropout=0.3).to(DEVICE)
    optimizer = torch.optim.Adam(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)

    best_val_f1 = -1.0
    best_state = None
    epochs_no_improve = 0

    for epoch in range(1, MAX_EPOCHS + 1):
        model.train()
        for x, y in train_loader:
            x, y = x.to(DEVICE), y.to(DEVICE)
            optimizer.zero_grad()
            loss = criterion(model(x), y)
            loss.backward()
            optimizer.step()

        _, _, _, _, val_f1, _, _, _ = evaluate_with_confusion(model, val_loader, criterion)

        if val_f1 > best_val_f1:
            best_val_f1 = val_f1
            best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
            epochs_no_improve = 0
        else:
            epochs_no_improve += 1
            if epochs_no_improve >= PATIENCE:
                break

    # Restore best, evaluate on test
    model.load_state_dict(best_state)
    test_loss, test_acc, test_p, test_r, test_f1, cm, preds, labels = \
        evaluate_with_confusion(model, test_loader, criterion)

    print(f"Fold {fold_idx}/{N_FOLDS}: "
          f"train_files={len(train_files):2d}  val={len(val_files):2d}  test={len(test_files):2d}  "
          f"| windows: tr={len(train_samples):3d} va={len(val_samples):2d} te={len(test_samples):2d}  "
          f"| best_val_F1={best_val_f1:.3f}  "
          f"test_F1={test_f1:.3f}  test_prec={test_p:.3f}  test_rec={test_r:.3f}  "
          f"test_acc={test_acc:.3f}  stopped@ep{epoch}")
    print(f"        confusion (rows=true, cols=pred):\n{cm}\n")

    fold_results.append({
        "fold": fold_idx,
        "test_loss": test_loss, "test_acc": test_acc,
        "test_p": test_p, "test_r": test_r, "test_f1": test_f1,
        "best_val_f1": best_val_f1,
        "test_files": [f.name for f in test_files],
        "stopped_epoch": epoch,
    })
    fold_models.append(best_state)


print("=" * 75)
print("5-FOLD CV SUMMARY")
print("=" * 75)
def stat(key):
    vals = np.array([r[key] for r in fold_results])
    return vals.mean(), vals.std()

for metric in ["test_acc", "test_p", "test_r", "test_f1"]:
    m, s = stat(metric)
    print(f"  {metric:12s}: {m:.3f} ± {s:.3f}   (per-fold: {[round(r[metric], 3) for r in fold_results]})")

winner_idx = int(np.argmax([r["test_f1"] for r in fold_results]))
winner = fold_results[winner_idx]
print(f"\nBest fold for deployment: Fold {winner['fold']} "
      f"(test F1 = {winner['test_f1']:.3f}, test files = {winner['test_files']})")

best_model_state = fold_models[winner_idx]
```

```python
KP_CONF_THRESHOLD = 0.3

def normalize_keypoints_v2(keypoints, bbox, conf_threshold=KP_CONF_THRESHOLD):
    kp = np.asarray(keypoints, dtype=np.float32).copy()
    x1, y1, x2, y2 = bbox

    cx = (x1 + x2) / 2.0
    cy = (y1 + y2) / 2.0
    half_w = max((x2 - x1) / 2.0, 1.0)
    half_h = max((y2 - y1) / 2.0, 1.0)

    kp[:, 0] = (kp[:, 0] - cx) / half_w
    kp[:, 1] = (kp[:, 1] - cy) / half_h

    bad = ~np.isfinite(kp).all(axis=1)
    kp[bad] = 0.0

    low_conf = kp[:, 2] < conf_threshold
    kp[low_conf] = 0.0

    return kp


def build_windows_from_file_v2(pkl_path, gt_path):
    with open(pkl_path, "rb") as f:
        data = pickle.load(f)
    gt = np.load(gt_path)

    tracklets = defaultdict(list)
    for frame_idx in sorted(data.keys()):
        for pid, person in data[frame_idx].items():
            bbox, keypoints = person[0], person[1]
            tracklets[pid].append((frame_idx, bbox, keypoints))

    samples = []
    for pid, entries in tracklets.items():
        if len(entries) < WINDOW:
            continue

        feats = np.stack([
            normalize_keypoints_v2(kp, bb).flatten()
            for _, bb, kp in entries
        ], axis=0)
        frame_indices = [fi for fi, _, _ in entries]

        for start in range(0, len(entries) - WINDOW + 1, STRIDE):
            window_feats = feats[start:start + WINDOW]
            window_frames = frame_indices[start:start + WINDOW]
            label = int(gt[window_frames].max())
            samples.append((window_feats.astype(np.float32), label))

    return samples


def windows_for_files_v2(file_list):
    out = []
    for pkl in file_list:
        gt = GT_DIR / (pkl.stem + ".npy")
        out.extend(build_windows_from_file_v2(pkl, gt))
    return out

sample_pkl = sorted(TEST_DIR.glob("*.pkl"))[0]
with open(sample_pkl, "rb") as f:
    sample_data = pickle.load(f)

total_kp = 0
zeroed_kp = 0
for frame_idx in sample_data:
    for pid, person in sample_data[frame_idx].items():
        bbox, kpts = person[0], person[1]
        kp_norm = normalize_keypoints_v2(kpts, bbox)
        total_kp += 17
        zeroed_kp += int((np.linalg.norm(kp_norm, axis=1) == 0).sum())

print(f"Sanity check on {sample_pkl.name}:")
print(f"  Total keypoints: {total_kp}")
print(f"  Zeroed (low-conf or NaN): {zeroed_kp}  ({100*zeroed_kp/total_kp:.1f}%)")
print(f"  Kept: {total_kp - zeroed_kp}\n")

SEED = 42
random.seed(SEED); np.random.seed(SEED)
torch.manual_seed(SEED); torch.cuda.manual_seed_all(SEED)

all_files = sorted(TEST_DIR.glob("*.pkl"))
kf = KFold(n_splits=N_FOLDS, shuffle=True, random_state=SEED)

fold_results_v2 = []
fold_models_v2  = []

print(f"5-fold CV with KP_CONF_THRESHOLD = {KP_CONF_THRESHOLD}\n")

for fold_idx, (train_idx, test_idx) in enumerate(kf.split(all_files), start=1):
    train_files = [all_files[i] for i in train_idx]
    test_files  = [all_files[i] for i in test_idx]

    rng = np.random.default_rng(SEED + fold_idx)
    perm = rng.permutation(len(train_files))
    n_val = max(3, len(train_files) // 6)
    val_files   = [train_files[i] for i in perm[:n_val]]
    train_files = [train_files[i] for i in perm[n_val:]]

    train_samples = windows_for_files_v2(train_files)
    val_samples   = windows_for_files_v2(val_files)
    test_samples  = windows_for_files_v2(test_files)

    train_loader = DataLoader(PoseLiftWindowDataset(train_samples), batch_size=BATCH_SIZE, shuffle=True)
    val_loader   = DataLoader(PoseLiftWindowDataset(val_samples),   batch_size=BATCH_SIZE, shuffle=False)
    test_loader  = DataLoader(PoseLiftWindowDataset(test_samples),  batch_size=BATCH_SIZE, shuffle=False)

    train_labels = np.array([y for _, y in train_samples])
    n_normal  = max(int((train_labels == 0).sum()), 1)
    n_anomaly = max(int((train_labels == 1).sum()), 1)
    w_normal  = (n_normal + n_anomaly) / (2.0 * n_normal)
    w_anomaly = (n_normal + n_anomaly) / (2.0 * n_anomaly)
    class_weights = torch.tensor([w_normal, w_anomaly], dtype=torch.float32, device=DEVICE)
    criterion = nn.CrossEntropyLoss(weight=class_weights)

    torch.manual_seed(SEED + fold_idx)
    model = ShoplifterLSTM(input_size=51, hidden_size=64, num_layers=2, dropout=0.3).to(DEVICE)
    optimizer = torch.optim.Adam(model.parameters(), lr=LR, weight_decay=WEIGHT_DECAY)

    best_val_f1 = -1.0
    best_state = None
    epochs_no_improve = 0

    for epoch in range(1, MAX_EPOCHS + 1):
        model.train()
        for x, y in train_loader:
            x, y = x.to(DEVICE), y.to(DEVICE)
            optimizer.zero_grad()
            loss = criterion(model(x), y)
            loss.backward()
            optimizer.step()

        _, _, _, _, val_f1, _, _, _ = evaluate_with_confusion(model, val_loader, criterion)

        if val_f1 > best_val_f1:
            best_val_f1 = val_f1
            best_state = {k: v.detach().cpu().clone() for k, v in model.state_dict().items()}
            epochs_no_improve = 0
        else:
            epochs_no_improve += 1
            if epochs_no_improve >= PATIENCE:
                break

    model.load_state_dict(best_state)
    test_loss, test_acc, test_p, test_r, test_f1, cm, _, _ = \
        evaluate_with_confusion(model, test_loader, criterion)

    print(f"Fold {fold_idx}/{N_FOLDS}: windows tr={len(train_samples):3d} va={len(val_samples):2d} te={len(test_samples):2d}  "
          f"| best_val_F1={best_val_f1:.3f}  test_F1={test_f1:.3f}  "
          f"prec={test_p:.3f}  rec={test_r:.3f}  acc={test_acc:.3f}  ep{epoch}")
    print(f"        confusion:\n{cm}\n")

    fold_results_v2.append({
        "fold": fold_idx, "test_acc": test_acc,
        "test_p": test_p, "test_r": test_r, "test_f1": test_f1,
        "best_val_f1": best_val_f1,
        "test_files": [f.name for f in test_files],
    })
    fold_models_v2.append(best_state)

print("=" * 75)
print("COMPARISON: v1 (no threshold) vs v2 (threshold=0.3)")
print("=" * 75)

def stat_v(results, key):
    vals = np.array([r[key] for r in results])
    return vals.mean(), vals.std()

print(f"{'Metric':<12} | {'v1 (mean ± std)':<22} | {'v2 (mean ± std)':<22} | Δ mean")
print("-" * 75)
for metric in ["test_acc", "test_p", "test_r", "test_f1"]:
    m1, s1 = stat_v(fold_results, metric)
    m2, s2 = stat_v(fold_results_v2, metric)
    delta = m2 - m1
    arrow = "↑" if delta > 0.01 else ("↓" if delta < -0.01 else "≈")
    print(f"{metric:<12} | {m1:.3f} ± {s1:.3f}        | {m2:.3f} ± {s2:.3f}        | {arrow} {delta:+.3f}")

winner_idx_v2 = int(np.argmax([r["test_f1"] for r in fold_results_v2]))
winner_v2 = fold_results_v2[winner_idx_v2]
print(f"\nv2 best fold: Fold {winner_v2['fold']} (test F1 = {winner_v2['test_f1']:.3f})")
```

```python
import json, torch, numpy as np, torch.nn as nn
from sklearn.metrics import (precision_score, recall_score, f1_score,
                             accuracy_score, confusion_matrix,
                             roc_curve, auc)
import matplotlib.pyplot as plt, matplotlib
matplotlib.use("Agg")
import os

EVAL_DIR = "/kaggle/working/evaluation"
os.makedirs(EVAL_DIR, exist_ok=True)

WINDOW = 30
STRIDE = 15
N_KP   = 17
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")

class ShopNet(nn.Module):
    def __init__(self, input_size=51, hidden_size=64, num_layers=2,
                 num_classes=2, dropout=0.3):
        super().__init__()
        self.lstm = nn.LSTM(input_size, hidden_size, num_layers,
                            batch_first=True, dropout=dropout)
        self.classifier = nn.Sequential(
            nn.Dropout(dropout),
            nn.Linear(hidden_size, num_classes),
        )

    def forward(self, x):
        _, (hn, _) = self.lstm(x)
        return self.classifier(hn[-1])

ckpt = torch.load(
    "/kaggle/input/datasets/nizarkabbaj/shoplifting-model/shoplifting_classifier.pt",
    map_location=DEVICE, weights_only=True)

cfg   = ckpt["model_config"]
model = ShopNet(**cfg).to(DEVICE)
model.load_state_dict(ckpt["state_dict"])
model.eval()
print(f"Model loaded on {DEVICE}")
print(f"Architecture: {cfg}")
```

```python
import pickle, pathlib
from collections import defaultdict
from sklearn.model_selection import KFold

DATA_DIR = pathlib.Path("/kaggle/input/datasets/nizarkabbaj/poselift-pickle")
TEST_DIR = DATA_DIR / "Test"
GT_DIR   = DATA_DIR / "GT"

test_files = sorted(TEST_DIR.glob("*.pkl"))  # 47 files
kf = KFold(n_splits=5, shuffle=True, random_state=42)
fold1_test_idx = list(kf.split(test_files))[0][1]
fold1_files    = [test_files[i] for i in fold1_test_idx]
print(f"Fold 1 test files: {len(fold1_files)}")

def normalize_keypoints(keypoints, bbox):
    kp = np.asarray(keypoints, dtype=np.float32).copy()
    x1, y1, x2, y2 = bbox
    cx = (x1 + x2) / 2.0
    cy = (y1 + y2) / 2.0
    half_w = max((x2 - x1) / 2.0, 1.0)
    half_h = max((y2 - y1) / 2.0, 1.0)
    kp[:, 0] = (kp[:, 0] - cx) / half_w
    kp[:, 1] = (kp[:, 1] - cy) / half_h
    bad = ~np.isfinite(kp).all(axis=1)
    kp[bad] = 0.0
    return kp


def extract_windows(pkl_path, gt_path):
    with open(pkl_path, "rb") as f:
        data = pickle.load(f)
    gt = np.load(gt_path)

    tracklets = defaultdict(list)
    for frame_idx in sorted(data.keys()):
        for pid, person in data[frame_idx].items():
            bbox, keypoints = person[0], person[1]
            tracklets[pid].append((frame_idx, bbox, keypoints))

    windows, labels = [], []
    for pid, entries in tracklets.items():
        if len(entries) < WINDOW:
            continue

        feats = np.stack([
            normalize_keypoints(kp, bb).flatten()
            for _, bb, kp in entries
        ], axis=0)
        frame_indices = [fi for fi, _, _ in entries]

        for start in range(0, len(entries) - WINDOW + 1, STRIDE):
            window_feats = feats[start:start + WINDOW]
            window_frames = frame_indices[start:start + WINDOW]
            label = int(gt[window_frames].max())
            windows.append(window_feats)
            labels.append(label)

    return np.array(windows, dtype=np.float32), np.array(labels, dtype=np.int64)
all_y_true, all_y_prob, all_y_pred = [], [], []

with torch.no_grad():
    for pkl_path in fold1_files:
        gt_path = GT_DIR / f"{pkl_path.stem}.npy"
        if not gt_path.exists():
            continue

        X, y = extract_windows(pkl_path, gt_path)
        if len(X) == 0:
            continue

        tensor = torch.tensor(X).to(DEVICE)
        logits = model(tensor)
        probs  = torch.softmax(logits, dim=1)[:, 1].cpu().numpy()
        preds  = (probs >= 0.5).astype(int)

        all_y_true.extend(y.tolist())
        all_y_prob.extend(probs.tolist())
        all_y_pred.extend(preds.tolist())

y_true = np.array(all_y_true)
y_prob = np.array(all_y_prob)
y_pred = np.array(all_y_pred)
print(f"Total windows: {len(y_true)}  |  positive rate: {y_true.mean():.2f}")
```

```python
prec  = precision_score(y_true, y_pred, zero_division=0)
rec   = recall_score(y_true, y_pred, zero_division=0)
f1    = f1_score(y_true, y_pred, zero_division=0)
acc   = accuracy_score(y_true, y_pred)
cm    = confusion_matrix(y_true, y_pred)
fpr, tpr, _ = roc_curve(y_true, y_prob)
roc_auc = auc(fpr, tpr)

tn, fp, fn, tp = cm.ravel()
per_class_acc_normal    = tn / (tn + fp) if (tn + fp) > 0 else 0
per_class_acc_anomaly   = tp / (tp + fn) if (tp + fn) > 0 else 0

print(f"Precision : {prec:.3f}")
print(f"Recall    : {rec:.3f}")
print(f"F1        : {f1:.3f}")
print(f"Accuracy  : {acc:.3f}")
print(f"AUC       : {roc_auc:.3f}")
print(f"Per-class accuracy — Normal: {per_class_acc_normal:.3f}  Anomaly: {per_class_acc_anomaly:.3f}")
print(f"\nConfusion matrix (TN FP / FN TP):\n{cm}")

fig, ax = plt.subplots(figsize=(5, 4))
im = ax.imshow(cm, interpolation="nearest", cmap="Blues")
plt.colorbar(im, ax=ax)
ax.set(xticks=[0,1], yticks=[0,1],
       xticklabels=["Normal","Anomaly"],
       yticklabels=["Normal","Anomaly"],
       xlabel="Predicted", ylabel="True",
       title="Confusion Matrix — Fold 1 (deployed model)")
for i in range(2):
    for j in range(2):
        ax.text(j, i, str(cm[i, j]), ha="center", va="center",
                color="white" if cm[i, j] > cm.max()/2 else "black", fontsize=14)
plt.tight_layout()
cm_path = f"{EVAL_DIR}/confusion_matrix.png"
plt.savefig(cm_path, dpi=150)
plt.show()
print(f"Saved → {cm_path}")

fig, ax = plt.subplots(figsize=(5, 4))
ax.plot(fpr, tpr, color="steelblue", lw=2, label=f"AUC = {roc_auc:.3f}")
ax.plot([0,1], [0,1], "k--", lw=1)
ax.set(xlabel="False Positive Rate", ylabel="True Positive Rate",
       title="ROC Curve — Fold 1 (deployed model)")
ax.legend(loc="lower right")
plt.tight_layout()
roc_path = f"{EVAL_DIR}/roc_curve.png"
plt.savefig(roc_path, dpi=150)
plt.show()
print(f"Saved → {roc_path}")
```

```python
metrics = {
    "fold": 1,
    "n_windows": int(len(y_true)),
    "precision": round(float(prec), 4),
    "recall":    round(float(rec),  4),
    "f1":        round(float(f1),   4),
    "accuracy":  round(float(acc),  4),
    "auc":       round(float(roc_auc), 4),
    "per_class_accuracy": {
        "normal":  round(float(per_class_acc_normal),  4),
        "anomaly": round(float(per_class_acc_anomaly), 4)
    },
    "confusion_matrix": cm.tolist()
}

metrics_path = f"{EVAL_DIR}/metrics.json"
with open(metrics_path, "w") as f:
    json.dump(metrics, f, indent=2)
print("Metrics saved →", metrics_path)
print(json.dumps(metrics, indent=2))
```
