```python
import os
import pickle
import numpy as np
import torch

# 1. Verify GPU
print("CUDA available:", torch.cuda.is_available())
print("GPU name:", torch.cuda.get_device_name(0) if torch.cuda.is_available() else "None")
print("PyTorch version:", torch.__version__)

# 2. Confirm file counts
base      = "/kaggle/input/datasets/nizarkabbaj/poselift-pickle"
train_dir = f"{base}/Train"
test_dir  = f"{base}/Test"
gt_dir    = f"{base}/GT"

train_files = sorted(os.listdir(train_dir))
test_files  = sorted(os.listdir(test_dir))
gt_files    = sorted(os.listdir(gt_dir))

print(f"\nTrain files : {len(train_files)}")
print(f"Test files  : {len(test_files)}")
print(f"GT files    : {len(gt_files)}")

# 3. Load one .pkl and inspect structure
sample_path = os.path.join(test_dir, test_files[0])
with open(sample_path, "rb") as f:
    sample = pickle.load(f)

print(f"\nLoaded      : {test_files[0]}")
print(f"Type        : {type(sample)}")
if isinstance(sample, dict):
    print(f"Keys        : {list(sample.keys())}")
elif isinstance(sample, list):
    print(f"Length      : {len(sample)}")
    print(f"First item  : {type(sample[0])}")
    if isinstance(sample[0], dict):
        print(f"First keys  : {list(sample[0].keys())}")
```

```python
# Inspect one frame from the loaded sample
frame_0 = sample[0]
print(f"Type of frame_0     : {type(frame_0)}")

if isinstance(frame_0, dict):
    print(f"Keys                : {list(frame_0.keys())}")
    for k, v in frame_0.items():
        print(f"  [{k}] type={type(v).__name__}", end="")
        if hasattr(v, 'shape'):
            print(f"  shape={v.shape}", end="")
        elif isinstance(v, (list, tuple)):
            print(f"  len={len(v)}", end="")
        print()
elif isinstance(frame_0, (list, tuple)):
    print(f"Length              : {len(frame_0)}")
    print(f"First element       : {frame_0[0]}")
```

```python
# Find first non-empty frame
for frame_id, frame_data in sample.items():
    if frame_data:
        print(f"First non-empty frame : {frame_id}")
        print(f"Type                  : {type(frame_data)}")
        if isinstance(frame_data, dict):
            print(f"Keys                  : {list(frame_data.keys())}")
            for k, v in frame_data.items():
                print(f"  [{k}] type={type(v).__name__}", end="")
                if hasattr(v, 'shape'):
                    print(f"  shape={v.shape}", end="")
                elif isinstance(v, (list, tuple)):
                    print(f"  len={len(v)}", end="")
                print()
        break
```

```python
# Inspect the person entry in frame 7
frame_7 = sample[7]
person = frame_7[3]  # person ID 3

print(f"person type    : {type(person)}")
print(f"person len     : {len(person)}")
for i, item in enumerate(person):
    print(f"\n  item[{i}] type={type(item).__name__}", end="")
    if hasattr(item, 'shape'):
        print(f"  shape={item.shape}")
        print(f"  values={item}")
    elif isinstance(item, (list, tuple)):
        print(f"  len={len(item)}")
        print(f"  values={item}")
    else:
        print(f"  value={item}")
```
